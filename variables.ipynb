# Variables

```python
x = 26
# x is a variable (a temporary place to store data)
print(x)
```

## Many values to multiple variables in a single line

```python
x, y, z = "Orange", "Banana", "Apple"
print(x, y, z, sep = '\n')
```

## One value to multiple Variables

```python
x = y = z = "Orange"
print(x, y, z, sep = '\n')
```

```python
y = "Apple"
y = x
print(x, y, z, sep = '\n')
y = "Pineapple"
print(x, y, z, sep = '\n')
```

## Unpacking a collection

```python
fruits = ["Orange", "Banana", "Apple"]
x, y, z = fruits
print(x, y, z, sep = '\n')
```

### mutability checking between list and tuple

#### List is mutable (Its values can be changed later)

```python
fruit_list = ["Orange", "Banana", "Apple"]
print(fruit_list)
fruit_list[2] = "Pineapple"
print(fruit_list)
```

#### Tuple is immutable but faster

```python
fruit_tuple = ("Orange", "Banana", "Apple")
print(fruit_tuple)
```

```python
fruit_tuple = ("Orange", "Banana", "Apple")
print(fruit_tuple)
fruit_tuple[3] = "Pineapple"
print(fruit_tuple)
```
