## Global Variables

```python
# Checking scope of variables
def test():
  n = 89
  print(n)

test()
```

```python
print(n)
```

Since the scope of n has expired it wont work outside the function.But it can be used everywhere if we declare it as a global variabvle outside the function

```python
m = "Awesome"
def myfunc():
  m = "Fantastic"
  print("Python is", m)
  print("Inside function")
myfunc()
print("Python is", m)
print("Outside Function")
```

```python
# Inside a function we can declare a global variable by using the global tag
def testfunc():
  global u
  u = 56
  print(u)

testfunc()
print(u)
```
