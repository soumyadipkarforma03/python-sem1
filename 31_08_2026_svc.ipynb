```python
fruits = [ 'apple', 'banana', 'cherry', 'kiwi']
print (fruits[3])
```

```python
newlist = []
for fruit in fruits:
  newlist.append(fruit)
print(newlist)
```

```python
newlist = []
for fruit in fruits:
  if 'a' in fruit:
    newlist.append(fruit)
print(newlist)
fruits.append('pineapple')
print(fruits)
```

```python
newlist = [fruit for fruit in fruits if "a" in fruit]
print(newlist)
print(fruits)
```

```python
new2list = [fruit.upper() if fruit != "banana" else "orange" for fruit in fruits]
print(new2list)
```

```python
new2list = [fruit.upper() for fruit in fruits]
print(new2list)
```
