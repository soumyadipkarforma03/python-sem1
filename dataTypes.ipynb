# **Data Types**

## Getting and Setting Type

```python
# Getting the type
x = 5
print(type(x))
y = "String"
print(type(y))
```

```python
x = b"Hello"
print(x)
print(type(x))
```

```python
# Setting specific data type
x = bytes(5)
print(x)
y = bool(0)
z = bool(5) # Anything greater than 0 is True
print(y, z)
x = set(("apple", "banana", "cherry"))
print(x)
x = dict(name="John", age=36)
print(x)
y = {"name" : "John", "age" : 36}
z = {"number" : "+10896754321", "dob" : "10.09.1999"}
p = [y, z]
print(p)
```
