# **Data Types**

## Getting and Setting Type

```python
# Getting the type
x = 5
print(type(x))
y = "String"
print(type(y))
```

```python
x = b"Hello"
print(x)
print(type(x))
```

```python
# Setting specific data type
x = bytes(5)
print(x)
y = bool(0) # False
z = bool(5) # Anything else than 0 is True
x = bool(-3) # Anything else than 0 is True
print(y, z, x)
x = set(("apple", "banana", "cherry"))
print(x)
x = dict(name="John", age=36)
print(x)
y = {"name" : "John", "age" : 36}
z = {"number" : "+10896754321", "dob" : "10.09.1999"}
p = [y, z]
print(p)
x = 1j
print(x)
print(type(x))
```

## Python Numbers

```python
# power of 10
x = 5e3
print(x)
y = 2.413e2
print(y)
print(type(y))
z = 3.14356e7
print(z)
print(type(z))
```

```python
# complex numbers
x = 3 + 5j
print(x)
```
